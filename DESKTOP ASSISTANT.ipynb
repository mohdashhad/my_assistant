```python
import tkinter as tk
from tkinter import scrolledtext
import pyttsx3
import speech_recognition as sr
import datetime
import os
import random
import webbrowser
import threading
import wikipedia  # Wikipedia API for lookups

# Initialize engine
engine = pyttsx3.init()
engine.setProperty('rate', 150)
voices = engine.getProperty('voices')
engine.setProperty('voice', voices[0].id)  # Change to voices[1].id if needed

def speak(text):
    output_box.insert(tk.END, f"Jarvis: {text}\n")
    output_box.see(tk.END)
    engine.say(text)
    engine.runAndWait()

# Response collections
greetings = ["Hey Ashhad!", "Hello there!", "Hi buddy!", "Nice to see you again!", "Hey!"]
feelings = ["I'm just a bunch of code, but I feel awesome with you!", "Always running smoothly!", "Feeling smart as ever!"]
jokes = [
    "Why did the computer go to art school? Because it had too many layers!",
    "I told my CPU a joke... but it still hasn’t responded!",
    "What do you call a smart jar? Jarvis, of course!"
]
moods = {
    "sad": "I'm here for you, Ashhad. Want me to play some music?",
    "bored": "Wanna hear a joke or open YouTube?",
    "happy": "That's great to hear! Spread the positivity!"
}

def run_jarvis(command):
    command = command.lower()

    if "hi ashhad" in command or "hello ashhad" in command:
        speak(random.choice(greetings))

    elif "how are you" in command:
        speak(random.choice(feelings))

    elif "tell me a joke" in command:
        speak(random.choice(jokes))

    elif "i am" in command:
        for mood in moods:
            if mood in command:
                speak(moods[mood])
                return
        speak("I'm not sure how to respond, but I'm here for you!")

    elif "your name" in command:
        speak("My name is Jarvis, your personal assistant.")

    elif "who created you" in command:
        speak("I was created by Ashhad, my brilliant master.")

    elif "time" in command:
        time = datetime.datetime.now().strftime("%I:%M %p")
        speak(f"The time is {time}")

    elif "date" in command:
        date = datetime.datetime.now().strftime("%A, %B %d, %Y")
        speak(f"Today is {date}")

    elif "open notepad" in command:
        os.system("notepad")
        speak("Opening Notepad")

    elif "open chrome" in command:
        path = "C:/Program Files/Google/Chrome/Application/chrome.exe"
        if os.path.exists(path):
            os.startfile(path)
            speak("Opening Chrome")
        else:
            speak("Chrome is not installed in the default location.")

    elif "play music" in command:
        music_folder = "C:/Users/Hp/Music"  # Change this to your music folder
        try:
            songs = [file for file in os.listdir(music_folder) if file.endswith(('.mp3', '.wav'))]
            if songs:
                song = random.choice(songs)
                os.startfile(os.path.join(music_folder, song))
                speak(f"Playing {song}")
            else:
                speak("No music files found in your music folder.")
        except:
            speak("Couldn't access the music folder.")

    elif "open youtube" in command:
        webbrowser.open("https://www.youtube.com")
        speak("Opening YouTube")

    elif "google" in command:
        search_query = command.replace("google", "").strip()
        if search_query:
            speak(f"Searching Google for {search_query}")
            webbrowser.open(f"https://www.google.com/search?q={search_query}")
        else:
            speak("Please specify what you'd like to search on Google.")

    elif "search wikipedia" in command:
        search_query = command.replace("search wikipedia", "").strip()
        if search_query:
            try:
                result = wikipedia.summary(search_query, sentences=2)
                speak(result)
            except wikipedia.exceptions.DisambiguationError as e:
                speak(f"Could you be more specific? There were multiple results: {e.options}")
            except wikipedia.exceptions.HTTPTimeoutError:
                speak("Sorry, there was an issue with the Wikipedia server.")
        else:
            speak("Please specify what you'd like to search on Wikipedia.")

    elif "exit" in command or "quit" in command:
        speak("Goodbye Ashhad! Have a great day!")
        root.quit()

    elif "help" in command:
        speak("You can say: time, date, open notepad, play music, google, weather, search wikipedia, or just chat with me!")

    else:
        fallback = [
            "Interesting... tell me more!",
            "Hmm, I'm learning every day.",
            "I didn't get that. Could you rephrase?",
            "You're fun to talk to!",
            "Wow! Sounds cool, Ashhad."
        ]
        speak(random.choice(fallback))

def submit_command():
    command = input_box.get()
    input_box.delete(0, tk.END)
    output_box.insert(tk.END, f"You: {command}\n")
    run_jarvis(command)

def voice_command():
    recognizer = sr.Recognizer()
    with sr.Microphone() as source:
        speak("Listening...")
        try:
            audio = recognizer.listen(source, timeout=5)
            command = recognizer.recognize_google(audio)
            output_box.insert(tk.END, f"You (voice): {command}\n")
            run_jarvis(command)
        except sr.WaitTimeoutError:
            speak("I didn't hear anything.")
        except sr.UnknownValueError:
            speak("Sorry, I couldn't understand that.")
        except sr.RequestError:
            speak("Check your internet connection.")

def start_listening():
    threading.Thread(target=voice_command).start()

# GUI Setup
root = tk.Tk()
root.title("Jarvis Desktop Assistant")
root.geometry("550x570")
root.configure(bg="#1e1e2f")

title = tk.Label(root, text="JARVIS", font=("Helvetica", 20, "bold"), fg="cyan", bg="#1e1e2f")
title.pack(pady=10)

output_box = scrolledtext.ScrolledText(root, width=65, height=18, wrap=tk.WORD, bg="#26263a", fg="white", font=("Courier", 10))
output_box.pack(pady=10)

frame = tk.Frame(root, bg="#1e1e2f")
frame.pack(pady=10)

input_box = tk.Entry(frame, width=40, font=("Helvetica", 12))
input_box.grid(row=0, column=0, padx=5)

submit_button = tk.Button(frame, text="Submit", command=submit_command, bg="cyan", fg="black", width=10)
submit_button.grid(row=0, column=1, padx=5)

mic_button = tk.Button(root, text="🎤 Speak", command=start_listening, bg="green", fg="white", font=("Helvetica", 10), width=20)
mic_button.pack(pady=10)

# Start up greeting
speak('Jarvis activated. Say "Hi Ashhad how can I assist you" to begin chatting!')

root.mainloop()
```
